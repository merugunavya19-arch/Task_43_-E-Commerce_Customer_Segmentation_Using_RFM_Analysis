## 1. Dataset and methodology
Source contains **541,909 rows and 8 columns**. Period: 2010-12-01 through 2011-12-09. It contains 5,268 duplicated rows and 135,080 missing CustomerID values. RFM reference date is the day after the last retained purchase (**2011-12-10**). Rows with missing customer IDs, cancelled invoices, duplicates, and non-positive quantity or price are excluded in the documented order.

**RFM definitions:** Recency = days since most recent purchase (lower is better); Frequency = distinct invoice count; Monetary = sum of line value (`Quantity × UnitPrice`). Scores 1–5 use quintiles; R is reversed so recent shoppers score higher. Example: R=5, F=4, M=5 gives combined score `545`.
